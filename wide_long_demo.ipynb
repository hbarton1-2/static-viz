```python
import pandas as pd
import altair as alt
from altair.datasets import data
```

```python
stocks = data.stocks()
stocks_wide = stocks.pivot(index="date", 
                           columns="symbol", 
                           values="price").add_prefix("price_").rename_axis(columns=None).reset_index()
stocks_wide
#has one row for each date
#one column for each stock symbol
```

```python
chart1 = alt.Chart(stocks_wide).mark_point().encode(
    x = 'date:T',
    y = 'price_GOOG:Q'
)

chart2 = alt.Chart(stocks_wide).mark_point(color = 'red').encode(
    x = 'date:T',
    y = 'price_AMZN:Q'
)

chart1 + chart2
```

```python
stocks_long = pd.wide_to_long(
    stocks_wide,
    stubnames='price', #prefixes -> value column
    i = 'date', #id column - one row per id in the wide data
    j = 'symbol', #category column 
    sep = '_',
    suffix=r"\w+" #non-numeric suffix
).reset_index() #automatically indexes by i and j

stocks_long
```

```python
chart = alt.Chart(stocks_long).mark_point().encode(
    x = 'date:T',
    y = 'price:Q',
    color = 'symbol:N'
)

chart
```
